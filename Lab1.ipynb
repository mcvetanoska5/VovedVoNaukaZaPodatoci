```python
url = "https://clevershop.mk/product-category/mobilni-laptopi-i-tableti/"
response = requests.get(url)
if response.status_code == 200:
    print("Successfully connected to the website!")
    soup = BeautifulSoup(response.content, 'html.parser')
    products = []
    product_containers = soup.select('.product-grid-item')
    for product in product_containers:
        title = product.select_one('.wd-entities-title').get_text(strip=True) if product.select_one('.wd-entities-title') else None
        prices = product.select('.woocommerce-Price-amount')
        regular_price = prices[0].get_text(strip=True) if prices else None
        discount_price = prices[1].get_text(strip=True) if len(prices) > 1 else None
        product_url = product.select_one('.wd-entities-title a')['href'] if product.select_one('.wd-entities-title a') else None
        add_to_cart_url = product.select_one('.add_to_cart_button')['href'] if product.select_one('.add_to_cart_button') else None
        products.append({
            'Title': title,
            'Regular Price': regular_price,
            'Discount Price': discount_price,
            'Product URL': product_url,
            'Add to Cart URL': add_to_cart_url
        })
    for product in products:
        print(product)
else:
    print(f"Failed to connect. Status code: {response.status_code}")
```

```python
base_url = "https://clevershop.mk/product-category/mobilni-laptopi-i-tableti"
def scrape_products(url):
    response = requests.get(url)
    if response.status_code != 200:
        print(f"Failed to retrieve page {url}")
        return []
    soup = BeautifulSoup(response.content, 'html.parser')
    products = []
    product_containers = soup.select('.product-grid-item')
    if not product_containers:
        return []
    for product in product_containers:
        title = product.select_one('.wd-entities-title').get_text(strip=True) if product.select_one('.wd-entities-title') else None
        prices = product.select('.woocommerce-Price-amount')
        regular_price = prices[0].get_text(strip=True) if prices else None
        discount_price = prices[1].get_text(strip=True) if len(prices) > 1 else None
        product_url = product.select_one('.wd-entities-title a')['href'] if product.select_one('.wd-entities-title a') else None
        add_to_cart_url = product.select_one('.add_to_cart_button')['href'] if product.select_one('.add_to_cart_button') else None
        products.append({
            'Title': title,
            'Regular Price': regular_price,
            'Discount Price': discount_price,
            'Product URL': product_url,
            'Add to Cart URL': add_to_cart_url
        })
    return products;
    all_products = [];
page = 1
while True:
    page_url = f"{base_url}/page/{page}/"
    page_products = scrape_products(page_url)
    if not page_products:
        break
    all_products.extend(page_products)
    print(f"Scraped page {page} successfully.")
    page += 1;
```

```python
df = pd.DataFrame(all_products)
print(df)
```

```python
df.to_csv('scraped_products.csv', index=False)
```
